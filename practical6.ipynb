```python
# Problem 1: BUGGY CODE
class Person:

    def __init__(self, name, id):
        self.name = name
        self.id = id

    def display(self):
        print("Name:", self.name)
        print("ID:", self.id)


class Emp(Person):

    def __init__(self, name, id):
        super().__init__(name, id)

    def Print(self):
        print("Emp class called")


Emp_details = Emp("Khevna", 103)

Emp_details.display()

Emp_details.Print()
```

```python
# Problem 2: Write a Python program to find Anagram Check.
def check_anagram(word1, word2):

    word1 = word1.lower()
    word2 = word2.lower()

    if sorted(word1) == sorted(word2):
        print("The given words are Anagrams.")
    else:
        print("The given words are not Anagrams.")


word1 = input("Enter first word: ")
word2 = input("Enter second word: ")

check_anagram(word1, word2)
```

```python
# Problem 3: Write a Python program to check for Armstrong Number.
num = int(input("Enter a number: "))

original_num = num

sum = 0

digits = len(str(num))

while num > 0:
    digit = num % 10
    sum = sum + digit ** digits
    num = num // 10

if sum == original_num:
    print(original_num, "is an Armstrong Number.")
else:
    print(original_num, "is not an Armstrong Number.")
```

```python
# Problem 4: Deal With Classes and Objects
class Vehicle:

    def __init__(self, max_speed, mileage):
        self.max_speed = max_speed
        self.mileage = mileage
        self.color = "white"
        self.seating_capacity = 0

    def assign_seating_capacity(self, capacity):
        self.seating_capacity = capacity

    def display_properties(self):
        print("Maximum Speed:", self.max_speed, "km/h")
        print("Mileage:", self.mileage, "km/l")
        print("Color:", self.color)
        print("Seating Capacity:", self.seating_capacity)
        print()


vehicle1 = Vehicle(200, 20)
vehicle1.assign_seating_capacity(5)

vehicle2 = Vehicle(180, 25)
vehicle2.assign_seating_capacity(4)


print("Vehicle 1 Details:")
vehicle1.display_properties()

print("Vehicle 2 Details:")
vehicle2.display_properties()
```
