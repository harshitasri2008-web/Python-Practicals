```python
#Part A - Dictioneries
# 1. Create an empty dictionary
inventory = {}
```

```python
# 2. Store the first product details in variables
ProductNo1_Name = "Mobile Phone"
ProductNo1_Quantity = 5
ProductNo1_Price = 20000
ProductNo1_releaseYear = 2020
```

```python
# 3. Add first product details to inventory
inventory["ProductNo1"] = {
    "Product Name": ProductNo1_Name,
    "Product Quantity": ProductNo1_Quantity,
    "Product Price": ProductNo1_Price,
    "Product Release Year": ProductNo1_releaseYear
}
```

```python
# 4. Store the second product details in variables
ProductNo2_Name = "Laptop"
ProductNo2_Quantity = 3
ProductNo2_Price = 50000
ProductNo2_releaseYear = 2022
```

```python
# 5. Add second product details to inventory
inventory["ProductNo2"] = {
    "Product Name": ProductNo2_Name,
    "Product Quantity": ProductNo2_Quantity,
    "Product Price": ProductNo2_Price,
    "Product Release Year": ProductNo2_releaseYear
}
```

```python
# 6. Display the products present in inventory
print("Products present in inventory:")
print(inventory)
```

```python
# 7. Check if release years are in inventory
print("\nChecking release years:")

if inventory["ProductNo1"]["Product Release Year"] == ProductNo1_releaseYear:
    print("ProductNo1_releaseYear is present in inventory.")
else:
    print("ProductNo1_releaseYear is not present in inventory.")

if inventory["ProductNo2"]["Product Release Year"] == ProductNo2_releaseYear:
    print("ProductNo2_releaseYear is present in inventory.")
else:
    print("ProductNo2_releaseYear is not present in inventory.")
```

```python
# 8. Delete release year of both products
del inventory["ProductNo1"]["Product Release Year"]
del inventory["ProductNo2"]["Product Release Year"]

print("\nInventory after deleting release years:")
print(inventory)
```

```python
#Part B - Tuples
# 1. Create a tuple of product prices
prices = (250, 300, 150, 400, 100, 350, 200)
```

```python
# 2. Find and print the highest price and the lowest price
highest_price = max(prices)
print("Highest price:", highest_price)
lowest_price = min(prices)
print("Lowest price:", lowest_price)
```

```python
# 3. Calculate and print the total sum
total = sum(prices)
print("Total sum:", total)
```

```python
# 4. Convert tuple into sorted list in ascending order
sorted_prices = sorted(prices)
print("Sorted list:", sorted_prices)
```

```python
# 5. Try to modify an element in the tuple
try:
    prices[0] = 500
except TypeError:
    print("Error: Tuple elements cannot be modified.")
```
